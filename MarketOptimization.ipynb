```python
#define factor functions 

def compute_momentum(close_prices, window=21):
    return close_prices.pct_change(periods=window)

def compute_volatility(close_prices, window=21):
    returns = close_prices.pct_change()
    return returns.rolling(window=window).std()

def compute_size(close, volume, window=21):
    size = close * volume
    return size.rolling(window=window).mean()
```

```python
tickers = data['Close'].columns
momentum_df = pd.DataFrame()
volatility_df = pd.DataFrame()
size_df = pd.DataFrame()


for ticker in tickers:
    close = data['Close'][ticker]
    volume = data['Volume'][ticker]
    
    momentum_df[ticker] = compute_momentum(close)
    volatility_df[ticker] = compute_volatility(close)
    size_df[ticker] = compute_size(close, volume)

valid_index = momentum_df.dropna().index
momentum_df = momentum_df.loc[valid_index]
volatility_df = volatility_df.loc[valid_index]
size_df = size_df.loc[valid_index]

# Then, further intersect with volatility and size valid indices
common_index = momentum_df.index \
    .intersection(volatility_df.dropna().index) \
    .intersection(size_df.dropna().index)

momentum_df = momentum_df.loc[common_index]
volatility_df = volatility_df.loc[common_index]
size_df = size_df.loc[common_index]
```

```python
print(close.isna().sum().sum())
print(volume.isna().sum().sum())
print(momentum_df.isna().sum().sum())
print(volatility_df.isna().sum().sum())
print(size_df.isna().sum().sum())

```

```python
# Creating a scoring system, to rank the stocks 
def rank(df, ascending=True):
    return df.rank(axis=1, method='first', ascending=ascending)


factor_scores = (rank(momentum_df, ascending=False)+
                rank(volatility_df, ascending=False)+
                rank(size_df, ascending=False))
```

```python
factor_scores.head()
```

```python
final_rankings = factor_scores.rank(axis=1, method='min')
```

```python
final_rankings.head()
```

```python
monthly_ranks = final_rankings.resample('M').last()

top_n = 5
top_stocks_per_month = monthly_ranks.apply(lambda row: row.nsmallest(top_n).index.tolist(), axis=1)
```

```python
# Create a DataFrame with weights
weights = pd.DataFrame(0, index=monthly_ranks.index, columns=monthly_ranks.columns)

for date, top_stocks in top_stocks_per_month.items():
    weights.loc[date, top_stocks] = 1 / top_n  # Equal weights
```

```python
daily_weights = weights.reindex(final_rankings.index)  # Expand to daily index
daily_weights = daily_weights.fillna(method='ffill')
```

```python
# Get daily returns
returns = data['Close'].pct_change().fillna(0)
```

```python
# Element-wise multiply and sum across stocks
portfolio_returns = (daily_weights * returns).sum(axis=1)
```

```python
cumulative_returns = (1 + portfolio_returns).cumprod()
cumulative_returns.plot(title='Factor-Based Portfolio Growth')
```
